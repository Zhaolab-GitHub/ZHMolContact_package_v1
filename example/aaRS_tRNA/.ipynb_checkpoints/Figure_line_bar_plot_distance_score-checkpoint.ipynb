```python
import pandas as pd
import os
import matplotlib.pyplot as plt
import seaborn as sns
plt.rcParams['font.family'] = 'Arial'
```

# Distance data

```python
distance_file = f'aaRSstRNA_C_A.pdb.distance.txt'
```

```python
distance_df = pd.read_csv(distance_file, sep='\t', header=None, names=['residue', 'nucleotide', 'dist'])
print(distance_df)

# 找出每个 nucleotide 的 dist 最小行的索引
idx = distance_df.groupby('nucleotide')['dist'].idxmin()

# 根据索引取行
min_dist_df = distance_df.loc[idx].reset_index(drop=True)

print(min_dist_df)

# 提取 nucleotide 列中的数字部分
min_dist_df['nuc_num'] = min_dist_df['nucleotide'].str.extract(r'C(\d+)').astype(int)

# 按数字排序
min_dist_df_sorted = min_dist_df.sort_values('nuc_num').reset_index(drop=True)

# 如果不需要辅助列，可以删除
min_dist_df_sorted = min_dist_df_sorted.drop(columns=['nuc_num'])

print(min_dist_df_sorted)
```

# Predicted data

```python
file = f'aaRSs_tRNA_pred.csv'
pred_df = pd.read_csv(file)
print(pred_df)
```

```python
df = pd.concat(
    [min_dist_df_sorted[['nucleotide']], 
     min_dist_df_sorted[['dist']], 
     pred_df[['pred_score']]],
    axis=1
)
# ---------- 标准化 dist 到 [0,1] ----------
df['dist_norm'] = (df['dist'] - df['dist'].min()) / (df['dist'].max() - df['dist'].min())
print(df)
```

```python
# 假设你的 DataFrame 名为 df
# df = pd.read_csv('your_file.csv')

# 提取 nucleotide 数字，用于排序
df['nuc_num'] = df['nucleotide'].str.extract(r'C(\d+)').astype(int)
df = df.sort_values('nuc_num').reset_index(drop=True)

# 创建图形和轴
fig, ax1 = plt.subplots(figsize=(6, 3))

# 绘制 dist 曲线（左 y 轴）
ax1.plot(df['nucleotide'], df['dist'], color='blue', marker='o', label='dist')
ax1.set_xlabel('Nucleotide')
ax1.set_ylabel('Distance', color='blue')
ax1.tick_params(axis='y', labelcolor='blue')
ax1.set_xticks(df['nucleotide'][::5])  # 每隔5个显示一次 X 轴标签
ax1.set_xticklabels(df['nucleotide'][::5], rotation=45)

# 创建右 y 轴
ax2 = ax1.twinx()
ax2.bar(df['nucleotide'], df['pred_score'], alpha=0.3, color='orange', label='pred_score')
ax2.set_ylabel('Pred Score', color='orange')
ax2.tick_params(axis='y', labelcolor='orange')

# 去掉上边和右边的边框
ax1.spines['top'].set_visible(False)
ax1.spines['right'].set_visible(False)
ax2.spines['top'].set_visible(False)
ax2.spines['right'].set_visible(False)

# 添加图例
ax1.legend(loc='upper left')
ax2.legend(loc='upper right')

# plt.title('Distance vs Pred Score per Nucleotide')
plt.tight_layout()
plt.show()
```
