# import

```python
import numpy as np
import pickle
from sklearn.manifold import TSNE
import matplotlib.pyplot as plt
import pandas as pd
import csv
import re
```

```python
# softmax 函数
def softmax(x):
    # 防止溢出：减去每行最大值
    e_x = np.exp(x - np.max(x, axis=1, keepdims=True))
    return e_x / np.sum(e_x, axis=1, keepdims=True)
```

# Import Data

```python
# 直接把你原始数据,读取文件
RNA_setattr_file = f'chimeraX cmd for color attr 2_cluster_RNA.txt'
RNA_setattr_df = pd.read_csv(RNA_setattr_file, sep='\t', names=[
    "setattr", "C_label", "resi", "BindInfo_label", "BindInfo_value",
    "create", "create_value", "type_label", "type_value"
])

print(RNA_setattr_df)

prot_setattr_file = f'chimeraX cmd for color attr 2_cluster_prot.txt'
prot_setattr_df = pd.read_csv(prot_setattr_file, sep='\t', names=[
    "setattr", "C_label", "resi", "BindInfo_label", "BindInfo_value",
    "create", "create_value", "type_label", "type_value"
])

print(prot_setattr_df)
```

# RNA protein

```python
with open('aaRSs_tRNA_RNA_K.pkl', 'rb') as f:
    RNA_s = pickle.load(f)

RNA_regions_df = pd.read_csv('tRNA_regions.csv')
# print(regions_df)
# 获取颜色标签
RNA_region_labels = RNA_regions_df["region"].values  # (节点数, )

with open('aaRSs_tRNA_prot_K.pkl', 'rb') as f:
    prot_s = pickle.load(f)

prot_regions_df = pd.read_csv('aaRSs_regions.csv')
# print(prot_regions_df)
# 获取颜色标签
prot_region_labels = prot_regions_df["region"].values  # (节点数, )
```

```python
K = [5, 10, 20]
```

```python
for idx, k in enumerate(K):
    
    RNA_s_K = RNA_s[idx].cpu().numpy()
    RNA_s_K = softmax(RNA_s_K)
    # print(RNA_s_K)
    clusters = np.arange(RNA_s_K.shape[1])
    # print(clusters)
    clusters_value = RNA_s_K.dot(clusters) / (RNA_s_K.shape[1]-1)
    # print(clusters_value)
        # 检查长度匹配
    if len(clusters_value) != len(RNA_setattr_df):
        raise ValueError(f"新数据长度 {len(clusters_value)} 与原表长度 {len(RNA_setattr_df)} 不匹配")
    
    # 替换 BindInfo_value 列
    RNA_setattr_df["BindInfo_value"] = clusters_value
    
    prot_s_K = prot_s[idx].cpu().numpy()
    prot_s_K = softmax(prot_s_K)
    # print(prot_s_K)
    clusters = np.arange(prot_s_K.shape[1])
    # print(clusters)
    clusters_value = prot_s_K.dot(clusters) / (prot_s_K.shape[1]-1)
    # print(clusters_value)
        # 检查长度匹配
    if len(clusters_value) != len(prot_setattr_df):
        raise ValueError(f"新数据长度 {len(clusters_value)} 与原表长度 {len(prot_setattr_df)} 不匹配")
    
    # 替换 BindInfo_value 列
    prot_setattr_df["BindInfo_value"] = clusters_value

    total_setattr_df = pd.concat([RNA_setattr_df, prot_setattr_df], axis=0)
    
    # CSV 文件名
    filename = f'chimeraX cmd for color attr 2_cluster_{k}.csv'
    
    # 写入 CSV
    total_setattr_df.to_csv(filename, header=None, index=None)
```
