```python
import numpy as np
import pickle
from sklearn.manifold import TSNE
import matplotlib.pyplot as plt
import pandas as pd
```

```python
with open('aaRSs_tRNA_RNA_K.pkl', 'rb') as f:
    RNA_s = pickle.load(f)

regions_df = pd.read_csv('tRNA_regions.csv')
print(regions_df)
# 获取颜色标签
region_labels = regions_df["region"].values  # (节点数, )
```

```python
RNA_s_K0 = RNA_s[0].cpu().numpy()
# RNA_s_K0 = (RNA_s_K0 - RNA_s_K0.min(axis=1, keepdims=True)) / \
#              (RNA_s_K0.max(axis=1, keepdims=True) - RNA_s_K0.min(axis=1, keepdims=True))
print(RNA_s_K0)
```

```python
import matplotlib.pyplot as plt
from sklearn.manifold import TSNE

# 假设 RNA_s_K0 已经是 numpy array
tsne_model = TSNE(n_components=2, random_state=2025, perplexity=5, n_iter=1000)
embedding = tsne_model.fit_transform(RNA_s_K0)  # (节点数, 2)

plt.figure(figsize=(8,8))
# 使用 region 赋色
scatter = plt.scatter(
    embedding[:, 0],
    embedding[:, 1],
    c=region_labels,
    cmap='tab10',      # 可换 'Set2', 'rainbow', 'viridis' 等
    s=50,
    alpha=0.9,
    edgecolors='k',
    linewidths=0.3
)

# 显示节点编号
for i, (x, y) in enumerate(embedding):
    plt.text(x+0.1, y+0.1, str(i), fontsize=8, color='red')  # +0.1 偏移避免重叠点

plt.title("t-SNE projection of node features with node IDs")
plt.xlabel("t-SNE 1")
plt.ylabel("t-SNE 2")
plt.show()
```

```python
import matplotlib.pyplot as plt
import umap
# 假设 RNA_s_K0 已经是 numpy array

umap_model = umap.UMAP(
    n_components=2,
    n_neighbors=5,   # 类似 t-SNE 的 perplexity
    min_dist=0.2,     # 控制聚类的紧密程度
    random_state=2025
)
embedding = umap_model.fit_transform(RNA_s_K0)  # (节点数, 2)
custom_colors = ['green', 'purple', 'blue', 'red', 'yellow']
plt.figure(figsize=(8, 8))
# 使用 region 赋色
scatter = plt.scatter(
    embedding[:, 0],
    embedding[:, 1],
    c=[custom_colors[label % len(custom_colors)] for label in region_labels],
    s=50,
    alpha=0.9,
    edgecolors='k',
    linewidths=0.3
)

# 显示节点编号
for i, (x, y) in enumerate(embedding):
    plt.text(x + 0.1, y + 0.1, str(i), fontsize=8, color='red')

plt.title("UMAP projection of node features with node IDs")
plt.xlabel("UMAP 1")
plt.ylabel("UMAP 2")
plt.show()
```
