```python
import pandas as pd
import os
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics import precision_score, recall_score, f1_score
plt.rcParams['font.family'] = 'Arial'
lsz = 20
tsz = 15
```

# Import Data

```python
file = f'aaRSs_tRNA_pred.csv'
pred_df = pd.read_csv(file)
print(pred_df)
```

```python
native_10A_file = f'aaRSstRNA_C_A_RNA_bindingsites_label_10A.txt'
native_10A_df = pd.read_csv(native_10A_file, sep='\t', header=None, names=['nucleotide', 'label_10A'])
print(native_10A_df)

native_8A_file = f'aaRSstRNA_C_A_RNA_bindingsites_label_8A.txt'
native_8A_df = pd.read_csv(native_8A_file, sep='\t', header=None, names=['nucleotide', 'label_8A'])
print(native_8A_df)
```

# plot

```python
import matplotlib.pyplot as plt

dist_cutoff = 10
df = pd.concat(
    [pred_df[['pred_label']], 
     native_10A_df[[f'label_{dist_cutoff}A']], 
    ] ,
    axis=1
)
print(df)
# 计算指标
y_true = df[f'label_{dist_cutoff}A']
y_pred = df['pred_label']
precision = precision_score(y_true, y_pred)
recall = recall_score(y_true, y_pred)
f1 = f1_score(y_true, y_pred)

print(f"Precision: {precision:.4f}")
print(f"Recall:    {recall:.4f}")
print(f"F1 Score:  {f1:.4f}")

# ---------------- 绘制横向空心柱状图 ----------------
metrics = ['Precision', 'Recall', 'F1 Score']
values = [precision, recall, f1]

fig, ax = plt.subplots(figsize=(5, 3.5))

# 横向空心柱状图（barh）
bars = ax.barh(
    metrics, 
    values, 
    facecolor='none',  # 内部透明
    edgecolor=['#ecabc2', '#bebaba', '#9d92c8'],  # 边框颜色
    linewidth=3,       # 边框宽度
    height=0.6         # 柱子厚度
)

# 在柱子右侧显示数值
for bar, val in zip(bars, values):
    ax.text(
        val + 0.02,                         # 在柱子右边稍微偏移
        bar.get_y() + bar.get_height() / 2, # 垂直居中
        f"{val:.3f}",
        ha='left', va='center',
        fontsize=tsz, fontweight='medium', color='#333333'
    )

# 美化坐标轴和标题
ax.set_xlim(0, 1.1)
ax.set_xlabel('Score', fontsize=lsz)
# ax.set_title('Precision, Recall, and F1-score', fontsize=12, fontweight='semibold', pad=10)

# 去掉顶部和右侧边框
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

# 优化刻度样式
ax.tick_params(axis='x', labelsize=tsz, length=0)
ax.tick_params(axis='y', labelsize=tsz)

# 添加轻微的左侧和底部边框
ax.spines['bottom'].set_linewidth(1.2)
ax.spines['left'].set_linewidth(1.2)

plt.tight_layout()
save_file = f'figure/Performance_{dist_cutoff}A_horizontal.png'
plt.savefig(save_file, dpi=600)
plt.show()
```
