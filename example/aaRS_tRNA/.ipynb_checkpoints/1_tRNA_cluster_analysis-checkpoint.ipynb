```python
import numpy as np
import pickle
from sklearn.manifold import TSNE
import matplotlib.pyplot as plt
import pandas as pd
```

```python
# softmax 函数
def softmax(x):
    # 防止溢出：减去每行最大值
    e_x = np.exp(x - np.max(x, axis=1, keepdims=True))
    return e_x / np.sum(e_x, axis=1, keepdims=True)
```

```python
with open('aaRSs_tRNA_RNA_K.pkl', 'rb') as f:
    RNA_s = pickle.load(f)

regions_df = pd.read_csv('tRNA_regions.csv')
# print(regions_df)
# 获取颜色标签
region_labels = regions_df["region"].values  # (节点数, )
```

```python
RNA_s_K0 = RNA_s[0].cpu().numpy()
# RNA_s_K0 = (RNA_s_K0 - RNA_s_K0.min(axis=1, keepdims=True)) / \
#              (RNA_s_K0.max(axis=1, keepdims=True) - RNA_s_K0.min(axis=1, keepdims=True))
RNA_s_K0 = softmax(RNA_s_K0)
print(RNA_s_K0)
clusters = np.arange(RNA_s_K0.shape[1])
print(clusters)
clusters_value = RNA_s_K0.dot(clusters) / (RNA_s_K0.shape[1]-1)
print(clusters_value)
```

```python
import matplotlib.pyplot as plt
from sklearn.manifold import TSNE

# 假设 RNA_s_K0 已经是 numpy array
tsne_model = TSNE(n_components=2, random_state=2025, perplexity=5, n_iter=1000)
embedding = tsne_model.fit_transform(RNA_s_K0)  # (节点数, 2)

plt.figure(figsize=(8,8))
# 使用 region 赋色
scatter = plt.scatter(
    embedding[:, 0],
    embedding[:, 1],
    c=[custom_colors[label % len(custom_colors)] for label in region_labels],
    s=50,
    alpha=0.9,
    edgecolors='k',
    linewidths=0.3
)

# 显示节点编号
for i, (x, y) in enumerate(embedding):
    plt.text(x+0.1, y+0.1, str(i), fontsize=8, color='red')  # +0.1 偏移避免重叠点

plt.title("t-SNE projection of node features with node IDs")
plt.xlabel("t-SNE 1")
plt.ylabel("t-SNE 2")
plt.show()
```

```python
import matplotlib.pyplot as plt
import umap
# 假设 RNA_s_K0 已经是 numpy array

umap_model = umap.UMAP(
    n_components=2,
    n_neighbors=10,   # 类似 t-SNE 的 perplexity
    min_dist=0.4,     # 控制聚类的紧密程度
    random_state=2025
)
embedding = umap_model.fit_transform(RNA_s_K0)  # (节点数, 2)
custom_colors = ['green', 'purple', 'blue', 'red', 'yellow']
plt.figure(figsize=(8, 8))
# 使用 region 赋色
scatter = plt.scatter(
    embedding[:, 0],
    embedding[:, 1],
    c=[custom_colors[label % len(custom_colors)] for label in region_labels],
    s=50,
    alpha=0.9,
    edgecolors='k',
    linewidths=0.3
)

# 显示节点编号
for i, (x, y) in enumerate(embedding):
    plt.text(x + 0.1, y + 0.1, str(i), fontsize=8, color='red')

plt.title("UMAP projection of node features with node IDs")
plt.xlabel("UMAP 1")
plt.ylabel("UMAP 2")
plt.show()
```

```python
# 基础参数
num_rows, num_cols = RNA_s_K0.shape
x = np.arange(num_rows)  # 每行一个 x 位置

# 自定义颜色（列对应的颜色）
colors = ['#FF6666', '#66CCFF', '#99CC33', '#FFCC33', '#CC99FF']

# 绘制堆叠柱状图
bottom = np.zeros(num_rows)
for i in range(num_cols):
    plt.bar(x, RNA_s_K0[:, i], bottom=bottom, color=colors[i], edgecolor='k', label=f'Col {i+1}')
    bottom += RNA_s_K0[:, i]  # 叠加上一层

# 图形美化
plt.xlabel('Row Index')
plt.ylabel('Value (stacked)')
plt.title('Stacked Bar Chart by Row')
plt.legend(title='Columns', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()
```
